
Write a Python program to calculate Speed, Distance, or Time based on user input.

Program Title:
Speed, Distance, and Time Calculator

Program Objective:
To develop a program that calculates Speed, Distance, or Time based on the user’s choice and input values, using the standard motion formulas.

Introduction:
This program helps users calculate one of the three motion parameters — Speed, Distance, or Time — when the other two are known.

The formulas used are:

Speed = Distance / Time

Distance = Speed × Time

Time = Distance / Speed

The program provides a menu-driven system where the user can choose what to calculate. It also includes error handling for invalid inputs, negative values, and division by zero.

Program Logic:
Start the Program

Display Menu Options:

Calculate Speed
Calculate Distance
Calculate Time
Exit
Ask the user to choose an option (1–4).

Based on the user’s choice:

If choice = 1 (Speed):

Input: Distance, Time

Check if inputs are positive and Time ≠ 0

Formula: Speed = Distance / Time

Display Speed

If choice = 2 (Distance):

Input: Speed, Time

Check if inputs are positive

Formula: Distance = Speed × Time

Display Distance

If choice = 3 (Time):

Input: Distance, Speed

Check if inputs are positive and Speed ≠ 0

Formula: Time = Distance / Speed

Display Time

If choice = 4 (Exit):

Display a goodbye message

Stop the program

Display an error message

Repeat the process until the user selects "Exit".

```python
# Speed, Distance, and Time Calculator

# Function to calculate speed
def calculate_speed(distance, time):
    # Formula: Speed = Distance divided by Time
    speed = distance / time
    return speed

# Function to calculate distance
def calculate_distance(speed, time):
    # Formula: Distance = Speed multiplied by Time
    distance = speed * time
    return distance

# Function to calculate time
def calculate_time(distance, speed):
    # Formula: Time = Distance divided by Speed
    time = distance / speed
    return time

# Start of the program
while True:
    # Show menu to the user
    print("\n--- Speed, Distance, and Time Calculator ---")
    print("1. Calculate Speed")
    print("2. Calculate Distance")
    print("3. Calculate Time")
    print("4. Exit")

    # Ask user to choose what they want to calculate
    choice = input("Enter your choice from 1 to 4: ")

    # If user chooses 1, calculate speed
    if choice == "1":
        # Ask user for distance and time
        distance_input = input("Enter distance: ")
        time_input = input("Enter time: ")

        try:
            # Convert inputs to numbers
            distance = float(distance_input)
            time = float(time_input)

            # Check for valid positive numbers
            if distance < 0 or time <= 0:
                print("Distance must be more than 0 and time must be more than 0")
            else:
                # Call the function to calculate speed
                speed = calculate_speed(distance, time)
                print("The Speed is", speed)
        except:
            # Handle if user enters something that is not a number
            print("Please enter numbers only")

    # If user chooses 2, calculate distance
    elif choice == "2":
        speed_input = input("Enter speed: ")
        time_input = input("Enter time: ")

        try:
            speed = float(speed_input)
            time = float(time_input)

            if speed < 0 or time < 0:
                print("Speed and time must be more than 0")
            else:
                distance = calculate_distance(speed, time)
                print("The Distance is", distance)
        except:
            print("Please enter numbers only")

    # If user chooses 3, calculate time
    elif choice == "3":
        distance_input = input("Enter distance: ")
        speed_input = input("Enter speed: ")

        try:
            distance = float(distance_input)
            speed = float(speed_input)

            if distance < 0 or speed <= 0:
                print("Distance must be more than 0 and speed must be more than 0")
            else:
                time = calculate_time(distance, speed)
                print("The Time is", time)
        except:
            print("Please enter numbers only")

    # If user chooses 4, exit the program
    elif choice == "4":
        print("Goodbye! Thank you for using the calculator.")
        break

    # If user enters invalid choice
    else:
        print("Invalid choice. Please choose 1, 2, 3, or 4")
```
